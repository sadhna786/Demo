# Comparing GPT, BERT, and T5: Architecture & Use Cases

| Model | Architecture | Use Cases |
|-------|--------------|-----------|
| **GPT** | Decoder-only transformer; unidirectional (left-to-right) language modeling. | Text generation, completion, chatbots, creative writing, code generation. |
| **BERT** | Encoder-only transformer; bidirectional context via masked language modeling. | Text classification, sentiment analysis, question answering, named entity recognition, embedding generation. |
| **T5** | Encoder-decoder transformer; reframes all NLP tasks as text-to-text. | Translation, summarization, question answering, classification, text generation, multi-task learning. |

**Key Differences:**
- GPT: Optimized for generating coherent text, using only the decoder stack.
- BERT: Designed for understanding and extracting information from text, using only the encoder stack.
- T5: Uses both encoder and decoder, making it flexible for any NLP task that can be cast as text-to-text.

#  Summary

GPT, BERT, and T5 are foundational transformer-based models in NLP, each with distinct architectures and strengths. GPT uses a decoder-only transformer for open-ended text generation, making it ideal for chatbots, story writing, and code completion. BERT employs an encoder-only transformer, leveraging bidirectional context for understanding tasks like classification, sentiment analysis, and question answering, but is not designed for text generation. T5 combines both encoder and decoder stacks, reframing all NLP tasks as text-to-text problems, which allows it to handle translation, summarization, classification, and more using a unified approach.

Running inference with GPT-2 is straightforward using Hugging Face Transformers. The code below demonstrates how to load the model, encode a prompt, and generate text. This process can be adapted for other models and tasks, highlighting the flexibility of transformer architectures in modern NLP.

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel
```

```python
# Load pre-trained GPT-2 model and tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
# Input prompt
prompt = "Artificial intelligence is transforming"
```

```python

# Encode input and generate output
inputs = tokenizer.encode(prompt, return_tensors="pt")
outputs = model.generate(inputs, max_length=50, num_return_sequences=1)
```

```python

# Decode and print result
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(generated_text)
```
