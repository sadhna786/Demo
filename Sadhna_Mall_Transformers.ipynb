```python
'''Self-Attention in Transformers
Role:
Self-attention enables transformers to weigh the importance of each token in a sequence relative to every other token. This allows the model to capture contextual relationships, regardless of their distance in the input.

Diagram:
Input: [The, cat, sat, on, the, mat]

      ┌─────────────┐
      │   Token 1   │
      └─────┬───────┘
            │
      ┌─────▼───────┐
      │   Token 2   │
      └─────┬───────┘
            │
      ┌─────▼───────┐
      │   Token 3   │
      └─────┬───────┘
            │
      ┌─────▼───────┐
      │   Token 4   │
      └─────┬───────┘
            │
      ┌─────▼───────┐
      │   Token 5   │
      └─────┬───────┘
            │
      ┌─────▼───────┐
      │   Token 6   │
      └─────────────┘

Each token attends to all other tokens:
[The] <--> [cat] <--> [sat] <--> [on] <--> [the] <--> [mat] 

Explanation:
Self-attention computes a weighted sum of all tokens for each token, allowing the model to focus on relevant words for understanding context.'''
```

```python
# Load BERT and print embeddings for sample text
from transformers import BertTokenizer, BertModel
import torch
```

```python
# Load pre-trained model and tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')
```

```python
# Sample text
text = "The cat sat on the mat."
```

```python
# Tokenize and get input tensors
inputs = tokenizer(text, return_tensors='pt')
```

```python
# Get embeddings from BERT
with torch.no_grad():
    outputs = model(**inputs)
    embeddings = outputs.last_hidden_state  # shape: [batch_size, seq_len, hidden_size]
```

```python
print("Embeddings shape:", embeddings.shape)
print("Embeddings for each token:\n", embeddings)
```
